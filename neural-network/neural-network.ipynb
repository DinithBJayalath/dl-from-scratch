```python
import numpy as np
```

```python
def sigmoid(x):
    return 1/(1 + np.exp(-x))
```

```python
class Neuron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias

    def feedforward(self, inputs):
        total = np.dot(self.weights, inputs) + self.bias
        return sigmoid(total)
```

```python
weights = np.array([0,1])
bias = 4
n = Neuron(weights, bias)
x = np.array([2,3])
print(n.feedforward(x))
```

```python
def mse_loss(y_ture, y_pred):
    return ((y_ture - y_pred) ** 2).mean()
```

```python
y_true = np.array([1, 0, 0, 1])
y_pred = np.array([0, 0, 0, 0])
print(mse_loss(y_true, y_pred))
```

```python
def sigmoid_derivative(input):
    fx = sigmoid(input)
    return fx * (1 - fx)
```

```python
class NeuralNetwork:
    def __init__(self):
        self.w1 = np.random.normal()
        self.w2 = np.random.normal()
        self.w3 = np.random.normal()
        self.w4 = np.random.normal()
        self.w5 = np.random.normal()
        self.w6 = np.random.normal()
        self.b1 = np.random.normal()
        self.b2 = np.random.normal()
        self.b3 = np.random.normal()

    def feedforward(self, x):
        h1 = sigmoid(self.w1 * x[0] + self.w2 * x[1] + self.b1)
        h2 = sigmoid(self.w3 * x[0] + self.w4 * x[1] + self.b2)
        out = sigmoid(self.w5 * h1 + self.w6 * h2 + self.b3)
        return out

    def train(self, data, y_true):
        learn_rate = 0.1
        epochs = 1000
        for epoch in range(epochs):
            for x, y_true_1 in zip(data, y_true):
                h1_sum = self.w1*x[0] + self.w2*x[1] + self.b1
                h1 = sigmoid(h1_sum)
                h2_sum = self.w3*x[0] + self.w4*x[1] + self.b2
                h2 = sigmoid(h2_sum)
                out_sum = self.w5*h1 + self.w6*h2 + self.b3
                output = sigmoid(out_sum)
                dL_dout = -2 * (y_true_1 - output)
                dout_dw5 = h1 * sigmoid_derivative(out_sum)
                dout_dw6 = h2 * sigmoid_derivative(out_sum)
                dout_db3 = sigmoid_derivative(out_sum)
                dout_dh1 = self.w5 * sigmoid_derivative(out_sum)
                dout_dh2 = self.w6 * sigmoid_derivative(out_sum)
                dh1_dw1 = x[0] * sigmoid_derivative(h1_sum)
                dh1_dw2 = x[1] * sigmoid_derivative(h1_sum)
                dh1_db1 = sigmoid_derivative(h1_sum)
                dh2_dw3 = x[0] * sigmoid_derivative(h2_sum)
                dh2_dw4 = x[1] * sigmoid_derivative(h2_sum)
                dh2_db2 = sigmoid_derivative(h2_sum)
                self.w1 -= learn_rate * dL_dout * dout_dh1 * dh1_dw1
                self.w2 -= learn_rate * dL_dout * dout_dh1 * dh1_dw2
                self.b1 -= learn_rate * dL_dout * dout_dh1 * dh1_db1
                self.w3 -= learn_rate * dL_dout * dout_dh2 * dh2_dw3
                self.w4 -= learn_rate * dL_dout * dout_dh2 * dh2_dw4
                self.b2 -= learn_rate * dL_dout * dout_dh2 * dh2_db2
                self.w5 -= learn_rate * dL_dout * dout_dw5
                self.w6 -= learn_rate * dL_dout * dout_dw6
                self.b3 -= learn_rate * dL_dout * dout_db3
            if epoch % 10 == 0:
                y_preds = np.apply_along_axis(self.feedforward, 1, data)
                loss = mse_loss(y_true, y_preds)
                print("Epoch %d loss: %.3f" % (epoch, loss))
```

```python
data = np.array([
  [-2, -1],
  [25, 6],
  [17, 4],
  [-15, -6],
])

y_true = np.array([
  1,
  0,
  0,
  1,
])

nn = NeuralNetwork()
nn.train(data, y_true)
```

```python
emily = np.array([-7, -3])
frank = np.array([20, 2])
print("Emily: %.3f" % nn.feedforward(emily))
print("Frank: %.3f" % nn.feedforward(frank))
```
